```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("bibliotecas importadas")
```

Leitura da base de dados

```python
df = pd.read_csv("/content/dataset_aula7_estatistica - dataset_aula7_estatistica.csv")
df.head()
```

```python
print("Dimensão da base:", df.shape)
print("\nColunas:")
print(df.columns)
```

Media
*   A média é a soma dos valores, dividida pela quantidade de elementos
A média indica o valor central medio do conjunto

```python
# Media temperatura
media_temperatura = df['temperatura_c'].mean()
print("media temperatura: ", media_temperatura)

# Media produção
media_producao = df["producao_dia"].mean()
print("media de produção ", media_producao)
```



*   Mediana

A mediana é o valor central quando os dados são colocados em ordem


```python
# Mediana da temperatura
mediana_temperatura = df ["temperatura_c"].median()
print("mediana da temperatura: ", mediana_temperatura)

#Mediana do tempo ciclo
mediana_tempo = df["tempo_ciclo_s"].median()
print("Mediana do tempo ciclo:", mediana_tempo)
```

# Moda
A moda é o valor que aparece com mais frequência


*Mode()

```python
# Moda do setor
moda_setor = df["setor"].mode()
print("moda do setor: ")
print(moda_setor)

# Moda da falha
moda_falha = df["falha"].mode()
print("\Moda da falha:")
print(moda_falha)
```

# Medida de dispersão
* amplitude
* variancia
* desvio padrão

```python
# Amplitude
amplitude_temperatura = df["temperatura_c"].max()- df["temperatura_c"].min()
print("Amplitude da temperatura:", amplitude_temperatura)
```

```python
# Variancia
variancia_temperatura = df["temperatura_c"].var()
print("Variancia da temperatura:", variancia_temperatura)
```

```python
# Desvio padrão temperatura
desvio_temperatura = df["temperatura_c"].std()
print("Desvio padrão da temperatura", desvio_temperatura)

#Desvio padrão produção
desvio_producao = df["producao_dia"].std()
print("Desvio padrão da produção: ", desvio_producao)
```

# Medidas de posição relativa

*   Quartis
*   Percentis

# Quartis
Dividem os dados em quatro partes Quartis principais

*   Q1-25%
*   Q2-50%
*   Q3-75%








```python
quartis_temperatura = df["temperatura_c"].quantile([0.25,0.50,0.75])
print("Quartis da temperatura: ")
print(quartis_temperatura)
```

# Percentis
Os percentis dividem os dados em cem partes

Exemplo
O percentis 90 vai indicar um valor abaixo do qual estão 90% dos dados

```python
# Percentis de 90%
p90_producao = np.percentile(df["producao_dia"],90)
print("Percentil 90 da produção: ",p90_producao)

# Percentis de 25%
p25_tempo = np.percentile(df["tempo_ciclo_s"],25)
print("Percentil 25 do tempo de ciclo: ",p25_tempo)
```

# Probabilidade Básica

 A probabilidade varia entre:


*   0 - impossivel
*   1 - certo



```python
# Probabilidade de ocorre falha
prob_falha = df["falha"].mean()
print("Probabilidade estimada de falha: ", prob_falha)
print(" Probabilidade em porcentagem: " , prob_falha* 100, "%")
```

```python
prob_temp_alta  = (df["temperatura_c"]>75).mean()
print("Probabilidade estimada de temperatura maior que 75ºC", prob_temp_alta)
print(f"Em porcentagem {prob_temp_alta * 100}%")
```

# Introdução a distribuição normal
principais características


*   Formato de sino
*   Concentração maior perto da média
*   Simetria em muitos casos



```python
# Distribuição aproximada da temperatura da base
sns.set_theme()

plt.figure(figsize=(8,5))
sns.histplot(data=df,
            x ="temperatura_c",
            bins=12,
            kde=True
            )
plt.title("Distribuição da temperatura")
plt.xlabel("Temperatura (Cº)")
plt.ylabel("Frequência")
plt.show()
```
